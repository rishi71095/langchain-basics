<small>

#### Why `init_chat_model` and what is it?

init_chat_model() is a LangChain factory function that creates a chat model instance for us

The reason we use init_chat_model() instead of:

```javascript
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="gpt-5-nano")
```

is because it gives **provider independence**.

Later you can switch to:

```javascript
model = init_chat_model(model="claude-sonnet-4")
```

or

```javascript
model = init_chat_model(model="gemini-2.5-pro")
```

without changing the rest of your code `(model-agnostic)`

This makes it easier to experiment with different model providers while keeping the same LangChain application logic.

#### Why `use_responses_api=True`

LangChain sends requests through OpenAI's newer Responses API instead of the older chat completions endpoint. It is designed to support:

- Tool calling
- Reasoning models
- Built-in web search
- Structured outputs

If we're building agents, workflows that will use tools, and applications that will use agents, then the Responses API is the recommended path because it is where OpenAI is adding new functionality.

</small>

## Customising your Model

````python
model = init_chat_model(
    model="gpt-5-nano",
    use_responses_api=True,
    # Kwargs passed to the model:
    temperature=1.0
)
````

````python

response = model.invoke("What's the capital of the Moon?")
print(response.text)
````

<small>
Args:

- `temperature`: higher temp means more creative output
- `max_output_tokens`: how long the output will be
- `timeout`: max time to wait for the model before cancelling the request
- `max_retries`: max amount of retries if that request fails

Passing chat history:

````python
from langchain.messages import AIMessage, HumanMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?"),
    AIMessage(content="The capital of the Moon is Luna City."),
    HumanMessage(content="Interesting, tell me more about Luna City")]}
)

print(response['messages'][-1].text)
````

````python
from langchain.messages import AIMessage, HumanMessage

chat_history = [
    HumanMessage(content="What's the capital of the Moon?"),
    AIMessage(content="The capital of the Moon is Luna City."),
    HumanMessage(content="Interesting, tell me more about Luna City in brief.")
    ]

response = agent.invoke(
    {"messages": chat_history }
)

print(response['messages'][-1].text)
````

## Streaming Output

Reduce latency by streaming tokens to the user:

````python
for token, metadata in agent.stream({"messages": [HumanMessage(content="Tell me all about Luna City, the capital of the Moon")]}, stream_mode="messages"):

    # token is a message chunk with token content
    # metadata contains which node produced the token
    
    if token.text:  # Check if there's actual content
        print(token.text, end="", flush=True)  # Print token
````

Metadata tells you where that streamed token came from in the LangGraph/LangChain execution. See metadata using the follwoing:

````python
for token, metadata in agent.stream({"messages": [HumanMessage(content="Hello")]}, stream_mode="messages"):
    pprint(metadata)
    break
````

## Basic prompting

````python
agent = create_agent(model="gpt-5-nano")

question = HumanMessage(content="What's the capital of the moon?")

response = agent.invoke(
    {"messages": [question]}
)

pprint(response['messages'][1].text)
````

## System prompt

````python
system_prompt = "You are a science fiction writer, create a capital city at the users request."

question = HumanMessage(content="What's the capital of the moon?")

scifi_agent = create_agent(
    model="gpt-5-nano",
    system_prompt=system_prompt
)

response = scifi_agent.invoke(
    {"messages": [question]}
)

print(response['messages'][1].text)
````

## Few-shot examples (in System prompt)


````python

system_prompt = """

You are a science fiction writer, create a space capital city at the users request.

User: What is the capital of mars?
Scifi Writer: Marsialis

User: What is the capital of Venus?
Scifi Writer: Venusovia

"""

question = HumanMessage(content="What's the capital of the moon?")

scifi_agent = create_agent(
    model="gpt-5-nano",
    system_prompt=system_prompt
)

response = scifi_agent.invoke(
    {"messages": [question]}
)

print(response['messages'][1].content)
````

# Streaming with prompts

Storing question and input separately:

````python
question = HumanMessage(
    content="Tell me all about Luna City, the capital of the Moon"
)

inputs = {
    "messages": [question]
}

for token, metadata in agent.stream(
    inputs,
    stream_mode="messages",
):
    if token.text:
        print(token.text, end="", flush=True)
````

## Structured prompts

````python
system_prompt = """

You are a science fiction writer, create a space capital city at the users request.

Please keep to the below structure.

Name: The name of the capital city

Location: Where it is based

Vibe: 2-3 words to describe its vibe

Economy: Main industries

"""

question = HumanMessage(content="What's the capital of the moon?")

scifi_agent = create_agent(
    model="gpt-5-nano",
    system_prompt=system_prompt
)

response = scifi_agent.invoke(
    {"messages": [question]}
)

print(response['messages'][1].content)
````

## Structured output

````python
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from pydantic import BaseModel

class CapitalInfo(BaseModel):
    name: str
    location: str
    vibe: str
    economy: str

agent = create_agent(
    model='gpt-5-nano',
    system_prompt="You are a science fiction writer, create a capital city at the users request.",
    response_format=CapitalInfo
)

question = HumanMessage(content="What is the capital of The Moon?")

response = agent.invoke(
    {"messages": [question]}
)

response["structured_response"]
````

````python
response["structured_response"].name
````

````python
capital_info = response["structured_response"]

capital_name = capital_info.name
capital_location = capital_info.location

print(f"{capital_name} is a city located at {capital_location}")
````
