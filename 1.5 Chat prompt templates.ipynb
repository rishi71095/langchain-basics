PromptTemplate creates a single text prompt by filling in variables, making it ideal for simple prompts and basic LLM interactions. ChatPromptTemplate creates a structured conversation with roles like system, human, and AI, which matches how modern chat models such as GPT-5 and Claude are designed to work.

For most modern AI applications, agents, and LangGraph workflows, ChatPromptTemplate is the preferred choice because it provides clearer instructions and supports multi-turn conversations.

```python
# importing PromptTemplate class from LangChain
from langchain_core.prompts import ChatPromptTemplate
# import the init_chat_model function
from langchain.chat_models import init_chat_model

# creating chat model instance
model = init_chat_model(
    model="gpt-5-nano",
    use_responses_api=True, # enabling the Responses API
    temperature=1.0
)

# user input
# topic = input("Enter a topic: ")

# creating a prompt template which contains a placeholder for the variable topic
prompt = ChatPromptTemplate.from_messages(
    [   
        ("system", "You are a funny comedian."),
        ("human", "Tell me a joke about {topic}")
    ]
)

# invoking the prompt template and passing in a dictionary containing a value for the topic variable
formatted_prompt = prompt.invoke(
    {"topic": "databases"}
)

# LangChain takes this dictionary and replaces the placeholder in the template with the value we provided
# At this point, the template becomes a fully formatted prompt that is ready to be sent to the language model

# invoking the language model with the formatted prompt
response = model.invoke(formatted_prompt)

print(response.text)
```

In LangChain, both the prompt and model are Runnables, meaning they can accept input and produce output.

```python
from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate

model = init_chat_model(
    model="gpt-5-nano",
    use_responses_api=True
)

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a witty comedian."),
        ("human", "Tell me a joke about {topic}.")
    ]
)

# The pipe operator is part of LangChain Expression Language (LCEL) and connects components together into a workflow.
# The output of the left component is passed as input to the right component.
# This creates a chain. A chain is simply multiple runnables connected together. The output of one step becomes the input of the next.
chain = prompt | model

# Dictionary Input
# Response object
response = chain.invoke(
    {"topic": "Vancouver"}
)

print(response.text)
```

```python
from langchain_core.prompts import ChatPromptTemplate
#StrOutputParser is a Runnable output parser in LangChain
#By default, the model returns an AIMessage object containing the response and metadata. StrOutputParser extracts just the text from that object and converts it into a plain Python string, making the output easier to work with.
from langchain_core.output_parsers import StrOutputParser

model = init_chat_model(
    model="gpt-5-nano",
    use_responses_api=True
)

prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in simple terms."
)

chain = prompt | model | StrOutputParser()

response = chain.invoke(
    {"topic": "LangGraph"}
)

# print(type(response))
print(response)
```
