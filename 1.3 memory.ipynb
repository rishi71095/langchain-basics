<small>

`Use an in-memory checkpoint storage system so the agent can save and restore conversation state.`

A checkpointer saves the graph's state after each step. That state can include:

- Conversation messages
- Tool outputs
- Intermediate agent state
- Any custom state in your graph

Think of checkpointer as a save game system for your agent.

The graph can reload previously saved state only when you invoke it again with the `same thread ID`.

`InMemorySaver` is the simplest checkpoint implementation. It stores checkpoints in Python memory (RAM).

If you restart Python, all saved conversations disappear.

Why do you need a thread ID? The checkpointer needs a key to know which conversation to restore.

Other option for local persistence:

`from langgraph.checkpoint.sqlite import SqliteSaver`

Stores checkpoints in a SQLite database file.

</small>

```python
from dotenv import load_dotenv, find_dotenv
from pprint import pprint
from langchain.agents import create_agent
```

```python
load_dotenv(find_dotenv(), override=True)
```

```python
from langgraph.checkpoint.memory import InMemorySaver  


agent = create_agent(
    "gpt-5-nano",
    checkpointer=InMemorySaver(),  
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Seán and my favourite colour is green")
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke(
    {"messages": [question]},
    config,  
)
```

```python
pprint(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")

response = agent.invoke(
    {"messages": [question]},
    config,  
)

pprint(response['messages'][-1].content)
```
